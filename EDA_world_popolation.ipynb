```python
# Count the number of unique values in each column, to see which columns have many different values and which have only a few (categories)
df.nunique()
```

```python
# Count the missing values in each column, to see where data is incomplete before analysing it
df.isnull().sum()
```

```python
# Count the rows that have at least one missing value, using the Rank column to count them
df[df.isnull().any(axis=1)]['Rank'].count()
```

```python
# Show the 10 countries with the largest 2022 population, by sorting from highest to lowest
df.sort_values(by = '2022 Population', ascending = False).head(10)
```

```python
# Show the 5 countries with the largest share of the world population, by sorting from highest to lowest percentage
df.sort_values(by='World Population Percentage', ascending = False).head()
```

```python
# Calculate the correlation between every pair of numeric columns, to see which columns move together
df.corr(numeric_only = True)
```

```python
# draw the correlations as a heatmap with the numbers written inside each square, so the strongest links stand out by colour
plt.rcParams['figure.figsize'] = (20, 7)
sns.heatmap(df.corr(numeric_only = True), annot = True)
plt.show
```

```python
# Group the countries by continent and calculate the average of each column, to compare continents with each other
df.groupby('Continent').mean(numeric_only = True)
```

```python
# Calculate the average of each column for every continent, then sort the continents from the highest to the lowest average 2022 population
df.groupby('Continent').mean(numeric_only = True).sort_values(by= '2022 Population', ascending = False)
```

```python
# Show only the rows where the continent contains the word "Africa", to look at the African countries 
df[df['Continent'].str.contains('Africa')]
```

```python
df[df['Continent']== 'Africa'].sort_values(by= '2022 Population', ascending = False).head(10)
# Show the 10 most populated African countries in 2022, sorting from the highest to the lowest population, top 10
```

```python
# For each continent, find the country with the largest 2022 population, and show only the continent, country and population columns
df.loc[df.groupby('Continent')['2022 Population'].idxmax(),['Continent','Country', '2022 Population']]
```
